Here is why environment variables are used instead of normal variables, followed by a step-by-step guide to setting up a safe .env file.
## Why Not a Normal Variable?
If you use a normal variable like gemini_key = "AIzaSy...", the secret key is written directly inside your Python file. This creates major security risks:

* Accidental Leaks: If you push your code to GitHub, GitLab, or share it with a friend, your secret key goes with it. Automated bots constantly scan public repositories for Gemini and OpenAI keys, and they will steal yours within seconds.
* Harder Collaborations: If a teammate wants to run your script, they have to open your code, erase your key, and type theirs.
* Environment Limits: Production servers (like AWS, Heroku, or Vercel) prefer injecting keys through the system environment rather than editing code files.

By using environment variables, your code stays completely separate from your secrets.
------------------------------
## How to Hide Your Key Safely Using a .env File
Follow these steps to move your key out of your code entirely.
## Step 1: Install the dotenv library
Open your terminal or command prompt and run this command to install the package that reads .env files:

pip install python-dotenv

## Step 2: Create the .env file
In the exact same folder where your Python script lives, create a new file named exactly .env (with the dot at the start and no file extension).
Open that file in your text editor and paste your key like this:

GEMINI_API_KEY=your_actual_api_key_here

Note: Do not use spaces around the = sign, and do not put quotes around your key.
## Step 3: Update your Python code
Now, update your Python script. You no longer need to type your actual key inside the code.

import osfrom dotenv import load_dotenv
# This automatically finds the .env file and loads the variables into os.environ
load_dotenv()
# Now Python can read it seamlesslyapi_key = os.environ.get("GEMINI_API_KEY")

print("Key loaded successfully!")

## Step 4: Protect it with .gitignore (Crucial)
If you use Git or GitHub, create another file in the same folder named .gitignore. Inside this file, type:

.env

This tells Git to completely ignore the .env file so it never gets uploaded to the internet.
